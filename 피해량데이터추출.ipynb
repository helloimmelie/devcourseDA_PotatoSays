```python
import pymysql
```

```python
sql = "SELECT cd.champion_name,  DATE_FORMAT(FROM_UNIXTIME(`timestamp`),'%H:%S') AS  `time`, AVG(pdd_to_champion) AS '물리피해량'," + \
"AVG(mdd_to_champion) AS '마법피해량', " + \
"AVG(time_enemy_controlled) AS '군중제어량' "+  \
"FROM LOL_data.champion_stat_per_timestamp AS spt " + \
"INNER JOIN LOL_data.champion_participant_id AS cpi ON spt.match_id = cpi.match_id AND spt.participant_id = cpi.participant_id " + \
"INNER JOIN champion_dict AS cd ON cpi.champion_id = cd.champion_id " + \
"WHERE champion_name IN ('가렌', '갈리오', '갱플랭크', '그라가스', '그레이브즈', '그웬', '나르', '나미', '나서스', '나피리', '노틸러스', '녹턴', '누누와 윌럼프', '니달리', '니코', '닐라', '다리우스', '다이애나', '드레이븐', '라이즈', '라칸', '람머스', '럭스', '럼블', '레나타 글라스크', '레넥톤', '레오나', '렉사이', '렐', '렝가', '루시안', '룰루', '르블랑', '리 신', '리븐', '리산드라', '릴리아', '마스터 이', '마오카이', '말자하', '말파이트', '모데카이저')" +\
" GROUP BY cd.champion_name,  DATE_FORMAT(FROM_UNIXTIME(`timestamp`),'%H:%S')"
```

```python
# 데이터베이스 연결 설정
db_config = {
    'host': 'localhost',
    'user': 'root',
    'password': 'pwd',
    'database': 'lol_data'
}
```

```python
# 배치 크기 설정
BATCH_SIZE = 50000

var_list = list()


```

```python
while True:
            # 데이터베이스 연결
    connection = pymysql.connect(**db_config)
    cursor = connection.cursor()
    cursor.execute(sql)
    with open("statResult_챔피언_그룹1.csv", "w", encoding = 'utf-8') as file:
        while True:
            # SQL 쿼리 실행
            batch_data = cursor.fetchmany(BATCH_SIZE)  # 배치 크기만큼 데이터 가져오기
            if len(batch_data) == 0:  # 더 이상 데이터가 없으면 종료
                break
        
            for row in batch_data:
                file.write(",".join(map(str, row)) + "\n")  # 데이터를 CSV 형식으로 저장
            # 또는 데이터를 다른 리스트나 파일에 저장할 수도 있음

    finally:
        # 연결 종료
    cursor.close()
    connection.close()

```
