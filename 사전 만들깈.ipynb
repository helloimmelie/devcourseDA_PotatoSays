```python
import json

with open('챔피언정보', 'r', encoding='utf-8') as file:
    champs = json.load(file)

with open('롤아이템템', 'r', encoding='utf-8') as file:
    items = json.load(file)
```

```python
champs_list = []
for x in champs['data']:
    champs_list.append((champs['data'][x]['key'], champs['data'][x]['name'], champs['data'][x]['tags']))

champs_list
```

```python
import pandas as pd
```

```python
cdf = pd.DataFrame(champs_list)
cdf.columns = ['key', 'name', 'tags']
cdf = cdf.set_index('key').sort_values('name')
cdf
```

```python
cdf.to_csv('챔프사전.csv', encoding='utf-8')
```

```python
items_list = []
for x in items['data']:
    items_list.append((x, items['data'][x]['name']))

items_list
```

```python
idf = pd.DataFrame(items_list)
idf.columns = ['key', 'name']
idf = idf.set_index('key')
idf
```

```python
idf.to_csv('아이템사전.csv', encoding='utf-8')
```

```python
import re
```

```python
def del_tags(x):
    x = re.sub(r"<[^>]+>", " ", x)
    x = re.sub(r"[^가-힣\s]", "", x)
    x = re.sub(r"\s+", " ", x)
    return x.strip()
```

```python
item_df['item_name'] = item_df['item_name'].map(del_tags)
item_df
```

```python
item_df.sort_values(by='item_name', key=lambda x: x.str.len(), ascending=False)
```

```python
item_df.sort_index().to_csv('아이템사전.csv', encoding='utf-8')
```
